# Validate Model Inference

1. Load predictions CSV
2. Compare predictions with actuals (if available)
3. Plot predicted vs actual
4. Calculate MAE and R2 on unseen data

```python
import pandas as pd
from sklearn.metrics import mean_absolute_error, r2_score
import matplotlib.pyplot as plt

# Load predictions
predictions = pd.read_csv("outputs/inference_results.csv")  # whatever path you saved

# Suppose your inference CSV has 'sales' (true) and 'predicted_sales' columns
y_true = predictions["sales"]
y_pred = predictions["predicted_sales"]

# Metrics
mae = mean_absolute_error(y_true, y_pred)
r2 = r2_score(y_true, y_pred)

print(f"Validation MAE: {mae}")
print(f"Validation R2: {r2}")

# Plot
plt.figure(figsize=(10,6))
plt.scatter(y_true, y_pred, alpha=0.5)
plt.xlabel("True Sales")
plt.ylabel("Predicted Sales")
plt.title("True vs Predicted Sales")
plt.grid(True)
plt.show()
```
